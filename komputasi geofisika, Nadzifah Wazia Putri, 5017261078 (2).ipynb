## Find a root a non-linnear equation using Bisection method

```python
import numpy as np
import math
```

```python
def bisection(a,b,tol,max_iteration):
    
        print('iter       a        b       c      f(a)     f(b)     f(c)     Err')
        print('--------------------------------------------------------------------')

        if func(a)*func(b)>0:
            print('Woi Error!')
            print('Change a & b!')
        else:
            for i in range(1, max_iteration):
                c = (a+b)/2
                Err= abs(a-b)
                
                print('{:2d}     {:3.4f}     {:3.4f}    {:3.4f}     {:3.4f}     {:3.4f}    {:3.4f}    {:3.4f}'
                .format(i, a, b, c, func(a), func(b), func(c), Err))
                
                if func(c)==0:
                    return c
                elif Err<tol:
                    break
                
                elif func(a)*func(c)<0:
                    b=c
                else:
                    a=c
            return c
        print('--------------------------------------------------------------------')
        print(f"The root of f(x)is", round(c,5))
```

## function
f(x) = xsin(x) -e^x, [-2,3], tol=0.001

```python
def func(x):
    return x**3 +2*x -2
```

```python
#no 1
c = bisection(-2, 3, 0.001, 50)

print('-------------------------------------------------------------------')
print(f"the root of f(x)is", round(c,5))
```
