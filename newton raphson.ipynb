# Root-Finding Method: Newton-Raphson

## $$ x_{new} = x_{old} - \frac{f(x-{old})}{f'(x_{old})} $$

```python
import numpy as np
import pandas as pd
```

# Define f(x)

## $f(x) = x sin x -e^x $

```python
def func(x):
    return x * np.sin(x) - np.exp(x)
```

# Define dereivative f'(x)

## $f'(x) = sinx + xcosx -e^x $

```python
def dfunc(x):
    return np.sin(x) + x*np.cos(x) - np.exp(x)
```

# Newton Raphson

```python
def newton(x0, tol):
    x_old = x0
    i = 1

    if dfunc(x0) == 0:
        print("This method doesn't work! The derivative is zero at the initial guess.")
        print("Please change the innitial guess x0.")
        return none

    x_new = x_old -func(x_old) / dfunc(x_old)
    Err = (abs(x_new - x_old) / abs(x_new))


    print('{:2d}  {:8.4f}   {:8.4f}   {:8.4f}   {:8.4f}   {:8.4f}'.format(
    i, x_old, func(x_old), dfunc(x_old), x_new, Err))

    while Err > tol:
        x_old = x_new
        x_new = x_old - func(x_old) / dfunc(x_old)
        Err = (abs(x_new - x_old))
        i= i+1

        print('{:2d}  {:8.4f}  {:8.4f}   {:8.4f}  {:8.4f}  {:8.4f}'.format(
           i, x_old, func(x_old), dfunc(x_old), x_new, Err))

        if func(x_new) == 0:
            return x_new

    return x_new
```

```python
print("iter   x_old   f(x_old)   f'(x_old)   x_new   Err(%)")
print("=======================================================")

root=newton(0.5,0.001)

print("=======================================================")
print(f"The Root of f(x) is ", round(root,5))
```

# Newton-Raphson Method with Data Frame

```python
def newton(x0, tol):
    rows = []
    i = 1


    while  True:
        if dfunc(x0) == 0:
            print("Turunan bernilai nola. Ganti x0,")
            return None

        x1 = x0 - func(x0) / dfunc(x0)
        err = abs(x1 - x0) / abs(x1) * 100 if x1 != 0 else np.inf

        rows.append([i, x0, func(x0), dfunc(x0), x1, err])

        if err <= tol or func(x1) == 0:
            break

        x0 = x1
        i += 1

    df = pd.DataFrame (rows, columns=[ "iterasi", "x_old", "f(x_old)", "f'(x_old)", "x_new", "Error (%)"])

    display(df.style.format({"x_old": "{:.5f}", "f(x_old)": "{:.5f}", "x_new": "{:.5f}", "Error (%)": "{:.5f}"}) 
            .hide(axis="index")
            .set_caption("Tabel Hasil Pencarian Akar dengan metode Newton-Raphson")
            .set_properties(**{"text-aling": "center"})
            .set_table_styles([{"selector": "caption", "props": [("font-size", "14px"), ("font-weight", "bold"),
                                                                 ("text-aling", "center"), ("padding", "8px")]},
                {"selector": "th", "props": [ ("background-color", "#2f5597"), ("color", "white"),
                    ("text-align", "center"), ("padding", "8px")]},
                {"selector": "td", "props": [("padding", "6px 12px")]},
                {"selector": "tbody tr:nth-child(even)", "props": [
                    ("background-color", "#f2f2f2")
                ]}
            ]))

    print(f"Root of f(x) ={x1:.5f}")
    return x1
```
